```python
weight=[0.1,0.1,-0.1]
toe=[8.5,9.5,9.9,9.0]
goal_pred=[0.1,0.0,0.0,0.1]
def sum_w(inp,weight):
  return sum(inp*weight[i] for i in range(len(weight)))
def mul_vec(inp,weight):
  return [sum_w(inp[i],weight) for i in range(len(inp))]
def MSE(pred,goal_pred):
  return (pred-goal_pred)**2
def delta(pred,goal_pred,alpha):
  return (pred-goal_pred)*alpha
alpha=0.001
epoch=200
tol=0.0001
wait=0
best_error=float('inf')
for iter in range(epoch):
  error=0
  pred=mul_vec(toe,weight)
  for i in range(len(toe)):
    error+=MSE(pred[i],goal_pred[i])
    for j in range(len(weight)):
      weight[j]-=delta(pred[i],goal_pred[i],alpha)*toe[i]
  if best_error>error+tol:
      best_error=error
      wait=0
  else:
      wait+=1
      if wait==10:
        print("Epoch:"+str(iter)+"Error:"+str(error))
        break
  if iter%10==0:
      print("Epoch:"+str(iter)+"error="+str(error))
```

```python
weight = [0.1, 0.1, -0.1]
toe = [8.5, 9.5, 9.9, 9.0]
goal_pred = [0.1, 0.0, 0.0, 0.1]

alpha = 0.01 # Немного увеличил, чтобы обучение шло быстрее
epoch = 200
tol = 0.0001
wait = 0
best_error = float('inf')

for iter in range(epoch):
    error = 0

    for i in range(len(toe)):
        # 1. Предсказание (скалярное произведение входного значения на веса)
        # В данной архитектуре мы считаем, что один вход toe[i] влияет на 3 выхода через 3 веса
        preds = [toe[i] * w for w in weight]

        # Считаем среднюю ошибку по весам для текущего примера
        # (Берем среднее отклонение, так как у нас 1 вход и 3 веса)
        example_error = sum((p - goal_pred[i])**2 for p in preds) / len(weight)
        error += example_error

        # 2. Обновление весов (Gradient Descent)
        for j in range(len(weight)):
            # Производная: (pred - goal) * input
            d = (preds[j] - goal_pred[i]) * toe[i]
            weight[j] -= alpha * d

    # Логика Early Stopping
    if error < best_error - tol:
        best_error = error
        wait = 0
    else:
        wait += 1

    if wait == 10:
        print(f"Early Stopping! Epoch: {iter} Error: {error:.6f}")
        break

    if iter % 10 == 0:
        print(f"Epoch: {iter} | Error: {error:.6f} | Weights: {[round(w, 4) for w in weight]}")

print("Final weights:", weight)
```
