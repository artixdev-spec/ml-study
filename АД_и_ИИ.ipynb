```python
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import pandas as pd


data = load_diabetes()
print("Data shape:", data.data.shape, "Target shape:", data.target.shape)

Xtr, Xt, Ytr, Yt = train_test_split(data.data, data.target, random_state=42)


model = tf.keras.models.Sequential()
model.add(tf.keras.layers.Input(shape=(10,)))
model.add(tf.keras.layers.Dense(1, activation='linear'))
model.summary()

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e0),
              loss='mse',
              metrics=['mse'])

es = tf.keras.callbacks.EarlyStopping(monitor='val_loss', mode='min', patience=3)

history = model.fit(Xtr, Ytr,
                   validation_split=0.2,
                   epochs=10000,
                   callbacks=[es],
                   verbose=1)

test_loss, test_mse = model.evaluate(Xt, Yt, verbose=0)
print(f"Test MSE: {test_mse}")

ym = model.predict(Xt)

print(model.weights)
plt.figure(figsize=(10, 6))
plt.scatter(Yt, ym)
plt.scatter(Yt, Yt)
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
trainx,testx,trainy,testy=train_test_split(data.data,data.target,random_state=42)
#Нормализация
scaler=StandardScaler()
trainx=scaler.fit_transform(trainx)
testx=scaler.transform(testx)
#3-х слойная нейронная сеть
model=tf.keras.Sequential([
    tf.keras.layers.Input(shape=(10,)),
    tf.keras.layers.Dense(50,activation="relu"),
    tf.keras.layers.Dense(30,activation="relu"),
    tf.keras.layers.Dense(1,activation="linear")
])
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
              loss="mse",
              metrics=["mse"])
history=model.fit(trainx,trainy,
                  validation_split=0.2,
                  verbose=1,
                  epochs=100)
test_loss,test_mse=model.evaluate(testx,testy,verbose=1)
print(f"MSE: {test_mse:.2f}")
y_pred=model.predict(testx,verbose=1).flatten()
r2=r2_score(testy,y_pred)
rmse=np.sqrt(test_mse)
print(f"RMSE",rmse)
print(f"R2 score:{r2:.4f}")
plt.figure(figsize=(12,5))
plt.subplot(1,3,1)
plt.title("обучение модели")
plt.xlabel("Эпоха")
plt.ylabel("MSE")
plt.plot(history.history['loss'],label="Train Loss")
plt.plot(history.history['val_loss'],label="Val Loss")
plt.legend()
plt.grid(True,alpha=0.3)
plt.subplot(1,3,2)
plt.title("Обуччение Модели(MSE)")
plt.xlabel("Эпоха",fontsize=12)
plt.ylabel("MSE",fontsize=12)
plt.plot(history.history["mse"],label="Train MSE")
plt.plot(history.history["val_mse"],label="Val MSE")
plt.grid(True,alpha=0.2)
plt.subplot(1, 3, 3)
plt.title(f"Предсказания (R²={r2:.4f})")
plt.xlabel("Фактические значения")
plt.ylabel("Предсказанные значения")
plt.scatter(testy, y_pred, alpha=0.6, label='Предсказания', color='blue', s=50)
min_val = min(testy.min(), y_pred.min())
max_val = max(testy.max(), y_pred.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=3, label='Идеальное')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
S=[]
for i in range(10000):
  indx=np.random.randint(0,X.shape[0],X.shape[0])
  S.append(np.percentile(X[indx],25))
np.percentile(S,2.5),np.percentile(S,97.5)
np.percentile(S,2.5)<X.mean()<np.percentile(S,97.5)
```

```python
bootstrap_means=[np.mean(X[np.random.randint(0,X.shape[0],X.shape[0])]) for i in range(1000)]
```

```python
bootstrap_median=[np.median(X[np.random.randint(0,X.shape[0],X.shape[0])]) for i in range(1000)]
```

```python
import numpy as np
import pandas as pd
from collections import deque

class MyDBSCAN:
    def __init__(self, eps=5.0, min_samples=5):
        self.eps = eps
        self.min_samples = min_samples
        self.labels_ = None

    def _get_neighbors(self, X, point_idx):
        """Найти всех соседей в радиусе eps"""
        neighbors = []
        for i in range(len(X)):
            # Евклидово расстояние
            dist = np.sqrt(np.sum((X[point_idx] - X[i]) ** 2))
            if dist <= self.eps:
                neighbors.append(i)
        return neighbors

    def fit(self, X):
        """Основной алгоритм DBSCAN"""
        n_samples = len(X)
        self.labels_ = np.full(n_samples, -1)  # -1 = шум
        cluster_id = 0
        visited = np.zeros(n_samples, dtype=bool)

        for point_idx in range(n_samples):
            if visited[point_idx]:
                continue

            visited[point_idx] = True
            neighbors = self._get_neighbors(X, point_idx)

            # Если недостаточно соседей — это шум
            if len(neighbors) < self.min_samples:
                self.labels_[point_idx] = -1
            else:
                # Начинаем новый кластер
                self._expand_cluster(X, point_idx, neighbors, cluster_id, visited)
                cluster_id += 1

        return self

    def _expand_cluster(self, X, point_idx, neighbors, cluster_id, visited):
        """Расширить кластер из core point"""
        self.labels_[point_idx] = cluster_id

        # Используем очередь для обхода
        queue = deque(neighbors)
        i = 0

        while i < len(queue):
            neighbor_idx = queue[i]

            if not visited[neighbor_idx]:
                visited[neighbor_idx] = True
                neighbor_neighbors = self._get_neighbors(X, neighbor_idx)

                # Если сосед тоже core point, добавляем его соседей
                if len(neighbor_neighbors) >= self.min_samples:
                    for nn in neighbor_neighbors:
                        if nn not in queue:
                            queue.append(nn)

            # Добавляем точку в кластер если ещё не добавлена
            if self.labels_[neighbor_idx] == -1:
                self.labels_[neighbor_idx] = cluster_id

            i += 1

# Использование
df = pd.read_csv("src2.dat", sep=r"\s+", header=None, names=["x", "y"])
X = df[['x', 'y']].values

# Создаём и обучаем модель
dbscan = MyDBSCAN(eps=5.0, min_samples=5)
dbscan.fit(X)

# Результаты
print(f"Уникальные метки: {set(dbscan.labels_)}")
print(f"Количество кластеров: {len(set(dbscan.labels_)) - (1 if -1 in dbscan.labels_ else 0)}")
print(f"Точек шума: {np.sum(dbscan.labels_ == -1)}")

# Визуализация
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 6))
plt.scatter(df['x'], df['y'], c=dbscan.labels_, cmap='viridis', alpha=0.5, s=10)
plt.colorbar(label='Cluster')
plt.show()
```

```python
with open('res_oval.dat','r') as f:
    data=[]
    for i in f:
        data.append([float(x) for x in i.split()])
data=np.array(data)
print(data[:10])
```
